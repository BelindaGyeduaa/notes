# File Processing

#### File Paths:
-   Absolute Paths - Beings at the root folder. Starts at the base of your computer.
    -   ex: `/Volumes/ExtHD/PerScholas/RTT-36/385/385.3/notes/notes.ipynb`
-   Relative paths - Path is relative to the current (working) directory
    -   ex: `./notes.ipynb`

#### File Path Symbols
-   `/` - divider between folders/files
-   `.` - currect directory
-   `..` - One directory up the chain

#### Opening Files
-   `open(file_path, mode)` - function

#### File modes
-   `r` - read - if file does not exists (FileNotFoundError)
-   `w` - write - if file doesnt exist, it creates a new file
-   `a` - append - if file doesnt exist, it creates a new file
-   `x` - exlusive creation - if file doesnt exist, it creates a new file
-   `b` - binary
-   `t` - text
-   `+` - read/write

#### Error Handling w/ Try: Except: Finally
-   `try:` - this block is code we want to run
-   `except:` - raise exceptions for errors we run into. Code to run if we get an error
-   `finally:` - code we run no matter what happens. errors or not

```python
try:
    f = open('data/my_file.txt', mode='r')

    print(f.read())

except Exception as e: # if an error, put in variable 'e' and print error to logs (terminal)
    print(f'❌ Error: {e}')

finally:
    f.close()
```

```python
# if opening file with 'with', all code goes into the with code block and the file auto closes at the end of the block
with open('data/my_file.txt', 'r') as f:
    print(f.read())
```

```python
def write_to_file(str, file_path):
    try:
        with open(file_path, 'w') as f:
            f.write(str + '\n')
            
        print('Success')
    except Exception as e:
        print(f'❌ Error: {e}')
```

```python
write_to_file('Belinda was here', 'data/my_file.txt')
```

```python
write_to_file('Darryl was here', 'data/my_file.txt')
```

```python
def append_to_file(str, file_path):
    try:
        with open(file_path, 'a') as f:
            f.write(str + '\n')
            
        print('Success')
    except Exception as e:
        print(f'❌ Error: {e}')
```

```python
append_to_file('Darryl was here', 'data/my_file.txt')
```
